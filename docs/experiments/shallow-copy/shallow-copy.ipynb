# Python 浅拷贝实验

运行两个单元前，先预测哪种操作会修改原列表 `a`。两段代码分别从同样的数据开始。

```python
a = [[1], [2]]
b = a.copy()
b[0].append(99)
print('a =', a)
print('b =', b)
print('a[0] is b[0]:', a[0] is b[0])
```

```python
a = [[1], [2]]
b = a.copy()
b[0] = [99]
print('a =', a)
print('b =', b)
print('a[0] is b[0]:', a[0] is b[0])
```

把第一个单元的 `b[0].append(99)` 改成 `b.append([99])`，然后重跑。预测这次改变了哪个列表。
